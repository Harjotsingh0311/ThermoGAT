```python
\
# ===========================================================================
# IMPORTS & GLOBAL SETUP
# ===========================================================================
import json
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 160)
warnings.filterwarnings("ignore")

# Use Path throughout (Windows-compatible). All paths are relative to the
# notebook's working directory, which should be the project root when run
# in VS Code (i.e. the folder that contains "results/").
BASE_DIR = Path("results")
FIG_DIR = BASE_DIR / "figures"
FIG_DIR.mkdir(parents=True, exist_ok=True)

# Running log of anything that went wrong or looked off, surfaced again in
# the final summary cell so nothing gets silently lost.
issues = []


def log_issue(msg):
    issues.append(msg)
    print(f"    [ISSUE] {msg}")


def safe_read_csv(path):
    '''Read a CSV, returning None (and logging) instead of raising.'''
    path = Path(path)
    if not path.exists():
        log_issue(f"Missing file: {path} — sections depending on it will be skipped.")
        return None
    try:
        return pd.read_csv(path)
    except Exception as e:
        log_issue(f"Could not read {path}: {e}")
        return None


def safe_read_json(path):
    '''Read a JSON file, returning None (and logging) instead of raising.'''
    path = Path(path)
    if not path.exists():
        log_issue(f"Missing file: {path} — sections depending on it will be skipped.")
        return None
    try:
        with open(path, "r") as f:
            return json.load(f)
    except Exception as e:
        log_issue(f"Could not read {path}: {e}")
        return None


def find_col(df, candidates):
    '''Case/underscore-insensitive column lookup. Returns the real column
    name from df, or None if nothing matches any candidate.'''
    if df is None:
        return None
    norm = {c.lower().replace(" ", "_"): c for c in df.columns}
    for cand in candidates:
        key = cand.lower().replace(" ", "_")
        if key in norm:
            return norm[key]
    return None


def get_key(d, candidates, default=None):
    '''Case-insensitive dict key lookup across a list of candidate names.'''
    if not isinstance(d, dict):
        return default
    norm = {k.lower(): k for k in d.keys()}
    for cand in candidates:
        key = cand.lower()
        if key in norm:
            return d[norm[key]]
    return default


def fmt_mean_std(mean, std, label="fold mean ± std"):
    if mean is None:
        return "n/a"
    if std is None or (isinstance(std, float) and np.isnan(std)):
        return f"{mean:.3f} (pooled value)"
    return f"{mean:.3f} ± {std:.3f} ({label})"


def roc_curve_np(y_true, y_score):
    '''Minimal numpy-only ROC curve (no sklearn — this notebook adds no
    new dependencies). Returns fpr, tpr sorted by ascending threshold.'''
    y_true = np.asarray(y_true).astype(int)
    y_score = np.asarray(y_score).astype(float)
    order = np.argsort(-y_score)
    y_true = y_true[order]
    y_score = y_score[order]
    n_pos = y_true.sum()
    n_neg = len(y_true) - n_pos
    if n_pos == 0 or n_neg == 0:
        return np.array([0, 1]), np.array([0, 1]), np.array([1, 0])
    tps = np.cumsum(y_true)
    fps = np.cumsum(1 - y_true)
    # keep only points where score changes (thresholds)
    distinct = np.where(np.diff(y_score) != 0)[0]
    idx = np.r_[distinct, len(y_score) - 1]
    tpr = np.r_[0, tps[idx] / n_pos]
    fpr = np.r_[0, fps[idx] / n_neg]
    thresh = np.r_[y_score[0] + 1, y_score[idx]]
    return fpr, tpr, thresh


def auc_trapz(fpr, tpr):
    order = np.argsort(fpr)
    x = np.asarray(fpr)[order]
    y = np.asarray(tpr)[order]
    # np.trapz was removed in numpy>=2.0 in favor of np.trapezoid; support both.
    trapz_fn = getattr(np, "trapezoid", None) or np.trapz
    return float(trapz_fn(y, x))


print("Setup complete. BASE_DIR =", BASE_DIR.resolve())
```

## 2. Master model-comparison table

One row per distinct model/config evaluated across the project. Only
**final, non-superseded** results are included (see the status table
above — Phase 11 and Phase 16 are intentionally excluded).

`roc_auc_or_range` is a string: a fold-mean ± std where cross-validated
fold-level results are available, otherwise a single pooled value — the
column itself, and the code below, states which for every row.


```python
\
# ===========================================================================
# 2. MASTER MODEL COMPARISON TABLE
# ===========================================================================
print("=" * 70)
print("SECTION 2 — MASTER MODEL COMPARISON TABLE")
print("=" * 70)

rows = []  # each: dict with model_name, category, roc_auc_or_range, balanced_accuracy, f1,
           # source_phase, source_file, _roc_sort (numeric, for sorting only)

CATEGORY_ORDER = ["primary", "tci_baseline", "structure_ablation", "bilateral", "final_headline"]


def add_from_fold_df(df, model_col, model_name_fixed, category, source_phase, source_file):
    '''Aggregate a fold-level dataframe (one row per fold) into a single
    master-table row with fold-mean ± std.'''
    if df is None:
        log_issue(f"Skipped '{model_name_fixed}' ({category}) — {source_file} not available.")
        return
    auc_col = find_col(df, ["roc_auc", "auc", "roc_auc_score"])
    bal_col = find_col(df, ["balanced_accuracy", "bal_acc", "balanced_acc"])
    f1_col = find_col(df, ["f1", "f1_score", "f1_macro"])
    if auc_col is None:
        log_issue(f"Skipped '{model_name_fixed}' ({category}) — no ROC AUC column found in {source_file}. "
                  f"Columns found: {list(df.columns)}")
        return
    sub = df
    if model_col and model_col in df.columns and model_name_fixed:
        sub = df[df[model_col].astype(str) == str(model_name_fixed)]
        if sub.empty:
            sub = df  # fall back to whole file if the exact name isn't found
    roc_mean, roc_std = sub[auc_col].mean(), sub[auc_col].std()
    bal_mean = sub[bal_col].mean() if bal_col else None
    f1_mean = sub[f1_col].mean() if f1_col else None
    rows.append({
        "model_name": model_name_fixed,
        "category": category,
        "roc_auc_or_range": fmt_mean_std(roc_mean, roc_std, "fold mean ± std"),
        "balanced_accuracy": round(bal_mean, 3) if bal_mean is not None else None,
        "f1": round(f1_mean, 3) if f1_mean is not None else None,
        "source_phase": source_phase,
        "source_file": source_file,
        "_roc_sort": roc_mean,
    })


def add_from_summary(summary, model_name, category, source_phase, source_file, fold_df=None, fold_file=None):
    '''Prefer a pre-computed *_summary.json (mean/std keys); fall back to
    aggregating the paired fold-level CSV if the summary is missing.'''
    if summary is not None:
        roc_mean = get_key(summary, ["roc_auc_mean", "auc_mean"])
        roc_std = get_key(summary, ["roc_auc_std", "auc_std"])
        bal_mean = get_key(summary, ["balanced_accuracy_mean", "bal_acc_mean"])
        f1_mean = get_key(summary, ["f1_mean"])
        if roc_mean is not None:
            rows.append({
                "model_name": model_name,
                "category": category,
                "roc_auc_or_range": fmt_mean_std(roc_mean, roc_std, "fold mean ± std"),
                "balanced_accuracy": round(bal_mean, 3) if bal_mean is not None else None,
                "f1": round(f1_mean, 3) if f1_mean is not None else None,
                "source_phase": source_phase,
                "source_file": source_file,
                "_roc_sort": roc_mean,
            })
            return
    # fall back to the fold-level CSV
    # fall back to the fold-level CSV. Use the caller-supplied real filename
    # when given (the "*_summary.json" -> "*_fold_metrics.csv" substitution
    # doesn't hold for every phase — e.g. Phase 24's fold-level file is
    # final_fusion_polish.csv, not final_fusion_polish_fold_metrics.csv).
    fallback_name = fold_file or source_file.replace("_summary.json", "_fold_metrics.csv")
    add_from_fold_df(fold_df, None, model_name, category, source_phase, fallback_name)


# --- Phase 13: DualBranchGAT primary (prefer cv_summary.json; else aggregate cv_fold_metrics.csv) ---
add_from_summary(
    data["cv_summary"], "DualBranchGAT (primary)", "primary", 13,
    "cv_summary.json", fold_df=data["cv_fold_metrics"], fold_file="cv_fold_metrics.csv",
)

# --- Phase 14: the 4 TCI baselines (single pooled value each, not cross-validated) ---
bc = data["baseline_comparison"]
if bc is not None:
    name_col = find_col(bc, ["model_name", "model", "name"])
    auc_col = find_col(bc, ["roc_auc", "auc"])
    bal_col = find_col(bc, ["balanced_accuracy", "bal_acc"])
    f1_col = find_col(bc, ["f1", "f1_score"])
    if name_col and auc_col:
        for _, r in bc.iterrows():
            rows.append({
                "model_name": str(r[name_col]),
                "category": "tci_baseline",
                "roc_auc_or_range": fmt_mean_std(r[auc_col], None, "pooled value"),
                "balanced_accuracy": round(r[bal_col], 3) if bal_col else None,
                "f1": round(r[f1_col], 3) if f1_col else None,
                "source_phase": 14,
                "source_file": "baseline_comparison.csv",
                "_roc_sort": r[auc_col],
            })
    else:
        log_issue("baseline_comparison.csv is missing model_name/roc_auc columns — TCI baselines skipped. "
                  f"Columns found: {list(bc.columns)}")
else:
    log_issue("Skipped TCI baselines (14) — baseline_comparison.csv not available.")

# --- Phase 17: 4 structure-ablation variants ---
# structure_ablation_results.csv is already ONE ROW PER VARIANT (pre-aggregated
# across folds), not fold-level data to be grouped-and-averaged. Column names
# use a "fold_<metric>_mean"/"fold_<metric>_std" convention (plus separate
# "pooled_<metric>" columns) rather than a bare "roc_auc" column — read the
# already-computed mean/std directly instead of trying to recompute them.
sa = data["structure_ablation_results"]
if sa is not None:
    variant_col = find_col(sa, ["variant", "variant_name", "model_name", "config", "topology"])
    auc_mean_col = find_col(sa, ["fold_roc_auc_mean", "roc_auc_fold_mean", "roc_auc_mean", "roc_auc", "auc"])
    auc_std_col = find_col(sa, ["fold_roc_auc_std", "roc_auc_fold_std", "roc_auc_std", "auc_std"])
    bal_col = find_col(sa, ["fold_balanced_accuracy_mean", "balanced_accuracy_fold_mean",
                             "balanced_accuracy_mean", "balanced_accuracy", "bal_acc"])
    f1_col = find_col(sa, ["fold_f1_mean", "f1_fold_mean", "f1_mean", "f1", "f1_score"])
    pooled_auc_col = find_col(sa, ["pooled_roc_auc", "roc_auc_pooled"])
    if variant_col and (auc_mean_col or pooled_auc_col):
        for _, r in sa.iterrows():
            if auc_mean_col is not None:
                roc_mean, roc_std = r[auc_mean_col], (r[auc_std_col] if auc_std_col else None)
                label = "fold mean ± std"
            else:
                roc_mean, roc_std = r[pooled_auc_col], None
                label = "pooled value"
            rows.append({
                "model_name": f"structure_ablation: {r[variant_col]}",
                "category": "structure_ablation",
                "roc_auc_or_range": fmt_mean_std(roc_mean, roc_std, label),
                "balanced_accuracy": round(r[bal_col], 3) if bal_col and pd.notna(r[bal_col]) else None,
                "f1": round(r[f1_col], 3) if f1_col and pd.notna(r[f1_col]) else None,
                "source_phase": 17,
                "source_file": "structure_ablation_results.csv",
                "_roc_sort": roc_mean,
            })
    else:
        log_issue("structure_ablation_results.csv is missing a variant/topology column or any "
                  f"roc_auc column — ablation rows skipped. Columns found: {list(sa.columns)}")
else:
    log_issue("Skipped structure ablation (17) — structure_ablation_results.csv not available.")

# --- Phase 23: bilateral graph (negative result, still reported) ---
# bilateral_graph_results.csv is a single already-aggregated summary row (not
# fold-level rows to average), with "<metric>_fold_mean"/"<metric>_fold_std"
# and separate "<metric>_pooled" columns — read those directly rather than
# routing through add_from_fold_df (which expects a bare "roc_auc" column
# it can groupby/mean/std itself).
bg = data["bilateral_graph_results"]
if bg is not None and len(bg) > 0:
    r = bg.iloc[0]
    auc_mean_col = find_col(bg, ["roc_auc_fold_mean", "fold_roc_auc_mean", "roc_auc_mean", "roc_auc"])
    auc_std_col = find_col(bg, ["roc_auc_fold_std", "fold_roc_auc_std", "roc_auc_std"])
    bal_col = find_col(bg, ["balanced_accuracy_fold_mean", "fold_balanced_accuracy_mean", "balanced_accuracy_mean"])
    f1_col = find_col(bg, ["f1_fold_mean", "fold_f1_mean", "f1_mean"])
    pooled_auc_col = find_col(bg, ["roc_auc_pooled", "pooled_roc_auc"])
    if auc_mean_col is not None:
        roc_mean, roc_std, label = r[auc_mean_col], (r[auc_std_col] if auc_std_col else None), "fold mean ± std"
    elif pooled_auc_col is not None:
        roc_mean, roc_std, label = r[pooled_auc_col], None, "pooled value"
    else:
        roc_mean, roc_std, label = None, None, None
    if roc_mean is not None:
        rows.append({
            "model_name": "BilateralGAT",
            "category": "bilateral",
            "roc_auc_or_range": fmt_mean_std(roc_mean, roc_std, label),
            "balanced_accuracy": round(r[bal_col], 3) if bal_col and pd.notna(r[bal_col]) else None,
            "f1": round(r[f1_col], 3) if f1_col and pd.notna(r[f1_col]) else None,
            "source_phase": 23,
            "source_file": "bilateral_graph_results.csv",
            "_roc_sort": roc_mean,
        })
    else:
        log_issue("Skipped 'BilateralGAT' (bilateral) — no recognizable roc_auc column found in "
                  f"bilateral_graph_results.csv. Columns found: {list(bg.columns)}")
else:
    log_issue("Skipped 'BilateralGAT' (bilateral) — bilateral_graph_results.csv not available.")

# --- Phase 24: final 3-way fusion (headline model) ---
add_from_summary(
    data["final_fusion_polish_summary"], "Final 3-way fusion (headline)", "final_headline", 24,
    "final_fusion_polish_summary.json", fold_df=data["final_fusion_polish"], fold_file="final_fusion_polish.csv",
)

# --- Assemble, sort, save ---
if rows:
    master = pd.DataFrame(rows)
    cat_rank = {c: i for i, c in enumerate(CATEGORY_ORDER)}
    master["_cat_rank"] = master["category"].map(cat_rank).fillna(len(CATEGORY_ORDER))
    master = master.sort_values(["_cat_rank", "_roc_sort"], ascending=[True, False])
    master = master.drop(columns=["_cat_rank", "_roc_sort"]).reset_index(drop=True)
    master = master[["model_name", "category", "roc_auc_or_range", "balanced_accuracy", "f1",
                      "source_phase", "source_file"]]

    print(master.to_string(index=False))
    out_path = BASE_DIR / "master_model_comparison.csv"
    master.to_csv(out_path, index=False)
    print(f"\nSaved -> {out_path}")
else:
    master = pd.DataFrame(columns=["model_name", "category", "roc_auc_or_range", "balanced_accuracy",
                                    "f1", "source_phase", "source_file"])
    out_path = BASE_DIR / "master_model_comparison.csv"
    master.to_csv(out_path, index=False)
    log_issue("Master table is EMPTY — no source files were available. "
              f"An empty (headers-only) {out_path} was still saved so the expected output exists.")
```

## 3. Figure 1 — main ROC comparison

Overlays ROC curves for TCI-alone, DualBranchGAT-alone, and the final
fusion model, using `final_model_predictions.csv` (`final_model_prob`,
`gat_prob`) and `baseline_pooled_predictions.csv` (`LR_worse_foot_tci`).
AUC is computed with a small numpy-only ROC implementation (no sklearn —
this notebook introduces no new dependencies) and shown in the legend.


```python
\
# ===========================================================================
# 3. FIGURE 1 — MAIN ROC COMPARISON
# ===========================================================================
print("=" * 70)
print("SECTION 3 — FIGURE 1: MAIN ROC COMPARISON")
print("=" * 70)

fmp = data["final_model_predictions"]
bpp = data["baseline_pooled_predictions"]
cvp = data["cv_pooled_predictions"]

fig1_path = FIG_DIR / "fig1_main_roc.png"

# final_model_predictions.csv (Phase 24) only carries the FINAL fused
# probability (final_model_prob), not a standalone GAT probability — the
# standalone GAT probability lives in cv_pooled_predictions.csv (Phase 13,
# column mc_mean_prob). All three files share subject_id, so merge on that
# rather than assuming any one file has every column needed.
if fmp is None or bpp is None or cvp is None:
    missing_files = [n for n, v in [("final_model_predictions.csv", fmp),
                                     ("baseline_pooled_predictions.csv", bpp),
                                     ("cv_pooled_predictions.csv", cvp)] if v is None]
    log_issue(f"Skipped Figure 1 — required file(s) missing: {missing_files}.")
else:
    subj_col_fmp = find_col(fmp, ["subject_id", "subject"])
    subj_col_bpp = find_col(bpp, ["subject_id", "subject"])
    subj_col_cvp = find_col(cvp, ["subject_id", "subject"])
    true_col_fmp = find_col(fmp, ["true_label", "label", "y_true"])
    fusion_col = find_col(fmp, ["final_model_prob", "fusion_prob", "fusion_b_prob"])
    tci_col = find_col(bpp, ["LR_worse_foot_tci", "tci_prob"])
    gat_col = find_col(cvp, ["mc_mean_prob", "gat_prob", "gat_probability"])

    missing = [n for n, v in [("subject_id (final_model_predictions)", subj_col_fmp),
                               ("subject_id (baseline_pooled_predictions)", subj_col_bpp),
                               ("subject_id (cv_pooled_predictions)", subj_col_cvp),
                               ("true_label", true_col_fmp), ("final_model_prob", fusion_col),
                               ("LR_worse_foot_tci", tci_col), ("mc_mean_prob (gat_prob)", gat_col)]
               if v is None]
    if missing:
        log_issue(f"Skipped Figure 1 — missing column(s): {missing}. "
                  f"final_model_predictions.csv columns: {list(fmp.columns)}; "
                  f"baseline_pooled_predictions.csv columns: {list(bpp.columns)}; "
                  f"cv_pooled_predictions.csv columns: {list(cvp.columns)}")
    else:
        merged = fmp[[subj_col_fmp, true_col_fmp, fusion_col]].rename(
            columns={subj_col_fmp: "subject_id", true_col_fmp: "true_label", fusion_col: "fusion_prob"})
        merged = merged.merge(
            bpp[[subj_col_bpp, tci_col]].rename(columns={subj_col_bpp: "subject_id", tci_col: "tci_prob"}),
            on="subject_id", how="inner")
        merged = merged.merge(
            cvp[[subj_col_cvp, gat_col]].rename(columns={subj_col_cvp: "subject_id", gat_col: "gat_prob"}),
            on="subject_id", how="inner")
        if merged.empty:
            log_issue("Skipped Figure 1 — merge on subject_id across the three prediction files "
                      "produced 0 rows; check that subject_id values match across files.")
        else:
            fig, ax = plt.subplots(figsize=(6, 6))
            curves = [
                ("TCI alone", merged["true_label"], merged["tci_prob"]),
                ("DualBranchGAT alone", merged["true_label"], merged["gat_prob"]),
                ("Final 3-way fusion", merged["true_label"], merged["fusion_prob"]),
            ]
            for label, y_true, y_score in curves:
                fpr, tpr, _ = roc_curve_np(y_true, y_score)
                auc = auc_trapz(fpr, tpr)
                ax.plot(fpr, tpr, label=f"{label} (AUC={auc:.3f})", linewidth=2)
            ax.plot([0, 1], [0, 1], linestyle="--", color="gray", linewidth=1, label="Chance")
            ax.set_xlabel("False Positive Rate")
            ax.set_ylabel("True Positive Rate")
            ax.set_title("Figure 1 — Main ROC Comparison")
            ax.legend(loc="lower right", fontsize=9)
            fig.tight_layout()
            fig.savefig(fig1_path, dpi=300)
            plt.show()
            print(f"Saved -> {fig1_path}  (n={len(merged)} subjects)")

```

## 8. Paper-section mapping

Maps every table/figure/finding to a suggested section of the paper, so
that writing up results is a matter of following this list rather than
re-deriving it.


```python
\
# ===========================================================================
# 8. PAPER-SECTION MAPPING
# ===========================================================================
print("=" * 70)
print("SECTION 8 — PAPER-SECTION MAPPING")
print("=" * 70)

# Pull the DeLong p-value dynamically from Phase 24's summary. The exact
# schema of final_fusion_polish_summary.json's DeLong section isn't fixed
# (it may be a flat key, a nested dict, or a list of per-comparison dicts),
# so search recursively for a p-value-shaped entry rather than assuming one
# exact top-level key name.
def find_delong_p_vs_tci(obj, path=()):
    # Recursively search a JSON-like structure for a p-value entry whose
    # context (containing dict keys, or a comparison/vs-style field) suggests
    # it is the winning-model-vs-TCI DeLong comparison. Returns the p-value
    # found and a short human-readable path for provenance, or (None, None).
    p_keys = ["p_value", "pval", "p", "delong_p", "delong_pvalue", "p_value_vs_tci"]
    if isinstance(obj, dict):
        for pk in p_keys:
            if pk in obj:
                context = " ".join(str(v) for v in obj.values() if isinstance(v, str)) + " " + " ".join(str(p) for p in path)
                if "tci" in context.lower() or any("tci" in str(p).lower() for p in path):
                    return obj[pk], "/".join(str(p) for p in path) + f"/{pk}"
        for k, v in obj.items():
            result = find_delong_p_vs_tci(v, path + (k,))
            if result[0] is not None:
                return result
    elif isinstance(obj, list):
        for i, item in enumerate(obj):
            result = find_delong_p_vs_tci(item, path + (i,))
            if result[0] is not None:
                return result
    return None, None


delong_p, delong_p_path = find_delong_p_vs_tci(data["final_fusion_polish_summary"])
if delong_p is None:
    # Fall back: grab ANY p-value-shaped entry anywhere in the file, so the
    # mapping cell still surfaces a candidate number for manual review
    # instead of silently reporting "not found" when a DeLong result does
    # exist under an unanticipated key name.
    def find_any_p(obj, path=()):
        p_keys = ["p_value", "pval", "p", "delong_p", "delong_pvalue"]
        if isinstance(obj, dict):
            for pk in p_keys:
                if pk in obj:
                    return obj[pk], "/".join(str(p) for p in path) + f"/{pk}"
            for k, v in obj.items():
                result = find_any_p(v, path + (k,))
                if result[0] is not None:
                    return result
        elif isinstance(obj, list):
            for i, item in enumerate(obj):
                result = find_any_p(item, path + (i,))
                if result[0] is not None:
                    return result
        return None, None
    delong_p, delong_p_path = find_any_p(data["final_fusion_polish_summary"])
    if delong_p is not None:
        log_issue(f"DeLong p-value for 'vs TCI' could not be confirmed by context — using best-guess "
                  f"value found at final_fusion_polish_summary.json:{delong_p_path}. Verify this is the "
                  f"winning-model-vs-TCI comparison before citing it in the paper.")

delong_note = (f"DeLong p={delong_p:.4f} vs TCI alone (source: final_fusion_polish_summary.json:{delong_p_path})"
               if delong_p is not None else
               "DeLong p-value vs TCI alone not found anywhere in final_fusion_polish_summary.json")
if delong_p is None:
    log_issue(f"Could not locate any DeLong p-value in final_fusion_polish_summary.json. "
              f"Top-level keys found: {list(data['final_fusion_polish_summary'].keys()) if isinstance(data['final_fusion_polish_summary'], dict) else type(data['final_fusion_polish_summary'])}")

demographics_note = (
    f"exact demographics-only AUC ({demographics_easy_auc:.3f}) vs. thermal-model AUC"
    if 'demographics_easy_auc' in dir() and demographics_easy_auc is not None
    else "demographics-only AUC not available (age_overlap_summary.json missing/incomplete)"
)

paper_section_mapping = [
    {"section": "Abstract / Results headline",
     "content": f"Final 3-way fusion metrics (Phase 24); {delong_note}."},
    {"section": "Methods / Architecture",
     "content": "Structure ablation findings (Phase 17) — robustness to edge choice as a design strength."},
    {"section": "Results, Table 1",
     "content": "master_model_comparison.csv"},
    {"section": "Results, Figure 1",
     "content": "fig1_main_roc.png"},
    {"section": "Results, Figure (secondary)",
     "content": "fig6_confusion_matrices.png, fig7_pr_curve.png"},
    {"section": "Results, robustness subsection",
     "content": "fig2_robustness.png, fig5_training_curve.png"},
    {"section": "Discussion, interpretability subsection",
     "content": "fig3_interpretability.png; MCA convergence with k-GCNDFU."},
    {"section": "Discussion, practical utility subsection",
     "content": "uncertainty-rejection finding (Phase 20); fig8_calibration.png."},
    {"section": "Discussion, negative-result subsection",
     "content": "Bilateral graph (Phase 23) — a deliberately-tested, honestly-reported null result."},
    {"section": "Limitations",
     "content": f"fig4_confound_check.png and the {demographics_note}."},
]

print(f"{'Section':<40s} | Content")
print("-" * 100)
for row in paper_section_mapping:
    print(f"{row['section']:<40s} | {row['content']}")

print("\n(Markdown table)\n")
print("| Section | Content |")
print("|---|---|")
for row in paper_section_mapping:
    print(f"| {row['section']} | {row['content']} |")

mapping_path = BASE_DIR / "paper_section_mapping.json"
with open(mapping_path, "w") as f:
    json.dump(paper_section_mapping, f, indent=2)
print(f"\nSaved -> {mapping_path}")
```
